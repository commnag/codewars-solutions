```python
# Two Sums

def two_sum(numbers, target):
    # [5,3,4,2,1], 6
    seen = {}
    final_result = []
    numbers_dict = dict(zip(range(len(numbers)), numbers))
    for index, number in numbers_dict.items():
        complement = target - number
        if complement in seen:
            final_result.append((index, seen[complement]))
        seen[number] = index
    return tuple(final_result)

two_sum([2,2,3], 4)
```

```python
#Digital Root

def digital_root(n):
    while n>10:
        n = sum([int(x) for x in list(str(n))])
    return n

digital_root(31425)
```

```python
# Bit Counting

def count_bits(n):
    return bin(n).count('1')

count_bits(24)
```
