```python
# Duplicate Encoder
def duplicate_encode(word):
    #your code here
    freq = {}
    word = word.lower()
    for c in word:
        if c in freq:
            freq[c] += 1
        else:
            freq[c] = 1
    result = ''
    for c in word:
        if freq[c] == 1:
            result += '('
        else:
            result += ')'

    return result

print(duplicate_encode('Success'))
```

```python
#Isograms
def is_isogram(string):
    #your code here
    string = string.lower()
    return False if len(set(string)) != len(string) else True
print(is_isogram('Success'))
```
