```python
# ROT13

def rot13(message):
    result = []
    for c in message:
        if c.isalpha():
            base = ord('a') if c.islower() else ord('A')
            c = chr((ord(c) - base + 13) % 26 + base)
        result.append(c)
    return ''.join(result)

rot13('This is my first ROT13 excercise!')
```
