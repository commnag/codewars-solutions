```python
# Write Number in Expanded Form

def expanded_form(num):
    num_str_lst = list(str(num))
    result = []
    for idx, i in enumerate(num_str_lst):
        sum = int(i) * 10**(len(num_str_lst) - int(idx) - 1)
        if sum > 0:
            result.append(str(sum))
    return ' + '.join(result)

expanded_form(70304)
```
