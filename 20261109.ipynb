```python
# Range Extraction

def solution(args):
    pieces = []
    i, n = 0, len(args)

    while i < n:
        # extend the run while numbers stay consecutive
        run_end = i
        while run_end + 1 < n and args[run_end + 1] == args[run_end] + 1:
            run_end += 1

        run_length = run_end - i + 1
        if run_length >= 3:
            pieces.append(f"{args[i]}-{args[run_end]}")
        else:
            pieces.extend(str(num) for num in args[i:run_end + 1])

        i = run_end + 1

    return ",".join(pieces)

solution([-3,-2,-1,2,10,15,16,18,19,20])
```

```python
# Is Square?

def is_square(n):
    return n >= 0 and int(n**0.5)**2 == n
```

```python
# Find the odd int

def find_it(seq):
    my_set = set(seq)
    for i in my_set:
        if seq.count(i) % 2 != 0:
            return i

find_it([20,1,-1,2,-2,3,3,5,5,1,2,4,20,4,-1,-2,5])
```

```python
# Playing with digits

def dig_pow(n, p):
    my_ls = list(str(n))
    tot = 0
    for i in my_ls:
        tot += int(i)**p
        p += 1

    return -1 if tot % n != 0 else tot // n

dig_pow(114, 3)
```

```python
# Sum of two lowest positive integer
def sum_two_smallest_numbers(numbers):
    return sum(sorted(numbers)[0:2])

sum_two_smallest_numbers([5, 8, 12, 18, 22])
```
